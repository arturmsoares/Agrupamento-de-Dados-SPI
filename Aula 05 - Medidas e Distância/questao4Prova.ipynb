```python
from sklearn.preprocessing import StandardScaler

import pandas as pd
from sklearn.cluster import KMeans

df = pd.DataFrame({
    "id_unidade": [101, 102, 103, 104, 105],
    "matriculas": [120, 140, 800, 850, 130],
    "valor_empenhado": [50000, 52000, 300000, 310000, 51000],
    "percentual_execucao": [0.90, 0.88, 0.40, 0.42, 0.89]
})

X = df[["matriculas", "valor_empenhado", "percentual_execucao"]]  # remove id

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

modelo = KMeans(n_clusters=2, random_state=42, n_init=10)
df["cluster"] = modelo.fit_predict(X_scaled)


print(df)
```
