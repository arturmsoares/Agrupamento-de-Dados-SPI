  Questão 2 — Efeito da normalização na proximidade.
Considere quatro unidades escolares descritas por dois atributos: número de matrículas e percentual de execução orçamentária. Deseja-se saber qual unidade é mais próxima de U1 usando distância Euclidiana.
Considere:

* Primeiro, os dados sem normalização. 
* Depois, os dados com normalização Min-Max em cada atributo.
Assinale a alternativa correta

Resposta: Sem normalização, a escala de matrículas (centenas) esmaga o percentual (0 a 1), fazendo U2 parecer próxima apenas por ter poucas matrículas. Após normalizar, o atributo de execução orçamentária ganha peso equivalente, revelando que U3 tem perfil realmente mais parecido com U1 (muitas matrículas e alta execução).

```python
import numpy as np
import pandas as pd

# Dados originais
dados = {
    "Unidade": ["U1", "U2", "U3", "U4"],
    "Matriculas": [100, 120, 500, 520],
    "Execucao": [0.80, 0.20, 0.75, 0.25]
}

df = pd.DataFrame(dados)

# ── 1. Distância Euclidiana SEM normalização ──────────────────────────────────
u1 = df[df["Unidade"] == "U1"][["Matriculas", "Execucao"]].values[0]

print("=== SEM NORMALIZAÇÃO ===")
for _, row in df[df["Unidade"] != "U1"].iterrows():
    ponto = row[["Matriculas", "Execucao"]].values
    dist = np.sqrt(np.sum((u1 - ponto) ** 2))
    print(f"Distância U1 → {row['Unidade']}: {dist:.4f}")

# ── 2. Normalização Min-Max ───────────────────────────────────────────────────
df["Matriculas_norm"] = (df["Matriculas"] - df["Matriculas"].min()) / \
                        (df["Matriculas"].max() - df["Matriculas"].min())

df["Execucao_norm"]   = (df["Execucao"] - df["Execucao"].min()) / \
                        (df["Execucao"].max() - df["Execucao"].min())

print("\n=== COM NORMALIZAÇÃO MIN-MAX ===")
print(df[["Unidade", "Matriculas_norm", "Execucao_norm"]].to_string(index=False))

# ── 3. Distância Euclidiana COM normalização ──────────────────────────────────
u1_norm = df[df["Unidade"] == "U1"][["Matriculas_norm", "Execucao_norm"]].values[0]

print("\n=== DISTÂNCIAS NORMALIZADAS ===")
for _, row in df[df["Unidade"] != "U1"].iterrows():
    ponto_norm = row[["Matriculas_norm", "Execucao_norm"]].values
    dist_norm = np.sqrt(np.sum((u1_norm - ponto_norm) ** 2))
    print(f"Distância U1 → {row['Unidade']}: {dist_norm:.4f}")
```
